```python
def predict_grocery_frequency(family_size):

    if family_size >= 6:
        return "once in two weeks"

    elif family_size >= 4:
        return "Monthly"

    elif family_size >= 2:
        return "2 months once"

    else:
        return "Monthly"
```

```python
predict_grocery_frequency(5)
```

```python
def predict_vegetable_frequency(family_size):

    if family_size >= 6:
        return "Daily"

    elif family_size >= 4:
        return "Twice a Week"

    elif family_size >= 2:
        return "Weekly"

    else:
        return "Biweekly"
```

```python
predict_vegetable_frequency(4)
```
