```python
vegetable_encoded = df["vegetables"].str.get_dummies(sep=";")

vegetable_encoded.columns = (
    "veg_" +
    vegetable_encoded.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_", regex=False)
)


df = pd.concat([df, vegetable_encoded], axis=1)

df.drop(columns=["vegetables"], inplace=True)

print(f"Vegetable Features Created: {vegetable_encoded.shape[1]}")
vegetable_encoded.head()
```

```python
print(df.columns)
```

```python
fruits_encoded = df["fruits"].str.get_dummies(sep=";")

fruits_encoded.columns = (
    "fruits_" +
    fruits_encoded.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_", regex=False)
)

df = pd.concat([df, fruits_encoded], axis=1)

df.drop(columns=["fruits"], inplace=True)
print(f"Fruits Features Created: {fruits_encoded.shape[1]}")
fruits_encoded.head()
```

```python
store_encoded = df["grocery_source"].str.get_dummies(sep=";")

store_encoded.columns = (
    "store_" +
    store_encoded.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_", regex=False)
)

df = pd.concat([df, store_encoded], axis=1)

df.drop(columns=["grocery_source"], inplace=True)
```

```python
dairy_encoded = df["dairy_products"].str.get_dummies(sep=";")

dairy_encoded.columns = (
    "dairy_" +
    dairy_encoded.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_", regex=False)
)

df = pd.concat([df, dairy_encoded], axis=1)

df.drop(columns=["dairy_products"], inplace=True)
```

```python
diet_encoded = pd.get_dummies(
    df["diet_type"],
    prefix="diet"
)

df = pd.concat([df, diet_encoded], axis=1)

df.drop(columns=["diet_type"], inplace=True)
```

```python
rice_encoded = pd.get_dummies(
    df["rice_type"],
    prefix="rice"
)

df = pd.concat([df, rice_encoded], axis=1)

df.drop(columns=["rice_type"], inplace=True)
```

```python
zone_encoded = pd.get_dummies(
    df["zone"],
    prefix="zone"
)

df = pd.concat([df, zone_encoded], axis=1)

df.drop(columns=["zone"], inplace=True)
```

```python
grocery_freq_map = {
    "Weekly": 1,
    "Twice a Week": 2,
    "Biweekly": 3,
    "Monthly": 4
}

df["grocery_frequency"] = df["grocery_frequency"].map(grocery_freq_map)
```

```python
veg_freq_map = {
    "Daily": 1,
    "Twice a Week": 2,
    "Weekly": 3,
    "Biweekly": 4,
    "Monthly": 5
}

df["vegetable_frequency"] = df["vegetable_frequency"].map(veg_freq_map)
```

```python
binary_columns = [col for col in df.columns if col.startswith(("veg_", "fruit_", "grain_", "dairy_", "store_", "diet_", "rice_", "zone_", "family_"))]

invalid = []

for col in binary_columns:
    values = set(df[col].unique())
    if not values.issubset({0, 1, True, False}):
        invalid.append((col, values))

if invalid:
    print(" Non-binary encoded columns found:")
    for col, values in invalid:
        print(f"{col}: {values}")
else:
    print(" All encoded columns are binary.")
```
